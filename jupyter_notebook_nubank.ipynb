```python
# Ref: state.country_id > country.country_id
# Ref: city.state_id > state.state_id

with engine.connect() as conn:   
    conn.execute(text("""ALTER TABLE country ALTER COLUMN country_id BIGINT NOT NULL"""))
    conn.execute(text("""ALTER TABLE state ALTER COLUMN state_id BIGINT NOT NULL"""))
    conn.execute(text("""ALTER TABLE city ALTER COLUMN city_id BIGINT NOT NULL"""))
    
    # Primary key
    conn.execute(text("""
        ALTER TABLE country
        ADD CONSTRAINT pk_country_id
        PRIMARY KEY (country_id)
    """))
    
    conn.execute(text("""
        ALTER TABLE state
        ADD CONSTRAINT pk_state_id
        PRIMARY KEY (state_id)
    """))
    
    conn.execute(text("""
        ALTER TABLE city
        ADD CONSTRAINT pk_city_id
        PRIMARY KEY (city_id)
    """))
    
    # Foreign key
    conn.execute(text("""
        ALTER TABLE state
        ADD CONSTRAINT fk_state_country
        FOREIGN KEY (country_id) REFERENCES country(country_id)
    """))

    conn.execute(text("""
        ALTER TABLE city
        ADD CONSTRAINT fk_city_state
        FOREIGN KEY (state_id) REFERENCES state(state_id)
    """))
    
    conn.commit()
```

```python
# Ref: customers.customer_city > city.city_id
# Ref: accounts.customer_id - customers.customer_id

with engine.connect() as conn:   
    conn.execute(text("""ALTER TABLE accounts ALTER COLUMN account_id BIGINT NOT NULL"""))
    conn.execute(text("""ALTER TABLE customers ALTER COLUMN customer_id BIGINT NOT NULL"""))
    
    # Primary key
    conn.execute(text("""
        ALTER TABLE accounts
        ADD CONSTRAINT pk_account_id
        PRIMARY KEY (account_id)
    """))
    
    conn.execute(text("""
        ALTER TABLE customers
        ADD CONSTRAINT pk_customer_id
        PRIMARY KEY (customer_id)
    """))
    
    # Foreign key
    conn.execute(text(""" 
        ALTER TABLE accounts
        ADD CONSTRAINT fk_account_customer
        FOREIGN KEY (customer_id) REFERENCES customers(customer_id)
    """)) 

    conn.execute(text("""
        ALTER TABLE customers
        ADD CONSTRAINT fk_customer_city
        FOREIGN KEY (customer_city) REFERENCES city(city_id)
    """))

    conn.commit()
```

```python
# Ref: transfer_ins.account_id > accounts.account_id
# Ref: transfer_outs.account_id > accounts.account_id
# Ref: pix_movements.account_id > accounts.account_id

with engine.connect() as conn:   
    conn.execute(text("""ALTER TABLE transfer_ins ALTER COLUMN id BIGINT NOT NULL"""))
    conn.execute(text("""ALTER TABLE transfer_outs ALTER COLUMN id BIGINT NOT NULL"""))
    conn.execute(text("""ALTER TABLE pix_movements ALTER COLUMN id BIGINT NOT NULL"""))
    
    # Primary key
    conn.execute(text("""
        ALTER TABLE transfer_ins
        ADD CONSTRAINT pk_transfer_ins_id
        PRIMARY KEY (id)
    """))
    
    conn.execute(text("""
        ALTER TABLE transfer_outs
        ADD CONSTRAINT pk_transfer_outs_id
        PRIMARY KEY (id)
    """))

    conn.execute(text("""
        ALTER TABLE pix_movements
        ADD CONSTRAINT pk_pix_movements_id
        PRIMARY KEY (id)
    """))
    
    # Foreign key
    conn.execute(text("""
        ALTER TABLE transfer_ins
        ADD CONSTRAINT fk_transfer_ins_account
        FOREIGN KEY (account_id) REFERENCES accounts(account_id)
    """))

    conn.execute(text("""
        ALTER TABLE transfer_outs
        ADD CONSTRAINT fk_transfer_outs_account
        FOREIGN KEY (account_id) REFERENCES accounts(account_id)
    """))

    conn.execute(text("""
        ALTER TABLE pix_movements
        ADD CONSTRAINT fk_pix_movements_account
        FOREIGN KEY (account_id) REFERENCES accounts(account_id)
    """))
    
    conn.commit()
```

```python
engine = create_engine(
    "mssql+pyodbc://localhost\\SQLEXPRESS/nubank_proposta?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes"
)

# ALTER AUTHORIZATION ON DATABASE::nubank_proposta TO sa; # to be consider when creating a SSMS diagram
```

```python
with engine.connect() as conn:
    # Delete column country_name, as it is redundant with the relationships between city, state and country tables and can lead to inconsistencies in the future
    conn.execute(text("""ALTER TABLE customers DROP COLUMN country_name""")) 
    
    # Rename column customer_city to city_id, to make it more clear that it is a foreign key to the city table and to follow naming conventions
    conn.execute(text("""
        EXEC sp_rename 'customers.customer_city', 'city_id', 'COLUMN'  
    """)) 
    
    # Changes to the customers table, to improve flexibility for different document types across countries, and to prevent data inconsistencies
    conn.execute(text("""
        EXEC sp_rename 'customers.cpf', 'document_number', 'COLUMN'  
    """))
   
    conn.execute(text("""ALTER TABLE customers ADD document_type VARCHAR(500)""")) 
    
    conn.execute(text("""UPDATE customers SET document_type = 'CPF' WHERE document_number IS NOT NULL""")) 

    # Delete tables/columns that assist in d_time, as values are already legible integers in the action_timestamp column and can be query by date functions
    conn.execute(text("""DROP TABLE d_week""")) 
    conn.execute(text("""DROP TABLE d_month""")) 
    conn.execute(text("""DROP TABLE d_weekday""")) 
    conn.execute(text("""DROP TABLE d_year""")) 
    
    conn.execute(text("""ALTER TABLE d_time DROP COLUMN week_id""")) 
    conn.execute(text("""ALTER TABLE d_time DROP COLUMN weekday_id""")) 
    conn.execute(text("""ALTER TABLE d_time DROP COLUMN month_id""")) 
    conn.execute(text("""ALTER TABLE d_time DROP COLUMN year_id""")) 
    
    conn.execute(text("""ALTER TABLE d_time ADD action_year INT""")) 
    conn.execute(text("""ALTER TABLE d_time ADD action_month INT""")) 
    conn.execute(text("""ALTER TABLE d_time ADD action_week INT""")) 
    conn.execute(text("""ALTER TABLE d_time ADD action_weekday INT""")) 
	
    conn.execute(text("""UPDATE d_time SET action_year = DATEPART(year,action_timestamp) """)) 
    conn.execute(text("""UPDATE d_time SET action_month = DATEPART(month,action_timestamp) """)) 
    conn.execute(text("""UPDATE d_time SET action_week = CAST(YEAR(action_timestamp) AS VARCHAR) + RIGHT('0' + CAST(DATEPART(ISO_WEEK, action_timestamp) - 1 AS VARCHAR), 2) """)) 
    conn.execute(text("""UPDATE d_time SET action_weekday = (DATEPART(weekday, action_timestamp) + @@DATEFIRST - 2 + 7) % 7  """)) 

    # New tables for transactions and products, enabling the addition of new types of movements in the future.
    conn.execute(text("""create table products (product_id BIGINT PRIMARY KEY, product_name VARCHAR(255), country_id BIGINT""")) 

    conn.execute(text("""INSERT INTO products VALUES ('1','Pix Movements', 1811589392032273152)""")) 
    conn.execute(text("""INSERT INTO products VALUES ('2','Transfer Movements', 1811589392032273152)""")) 
        
    conn.execute(text("""create table transactions_type (
					  transaction_type_id BIGINT PRIMARY KEY, 
					  description VARCHAR(255)
                      )
					  """)) 

    conn.execute(text("""INSERT INTO transactions_type VALUES ('11','pix_in')""")) 
    conn.execute(text("""INSERT INTO transactions_type VALUES ('12','pix_out')""")) 
    conn.execute(text("""INSERT INTO transactions_type VALUES ('21','transfer_in')""")) 
    conn.execute(text("""INSERT INTO transactions_type VALUES ('22','transfer_out')""")) 
    
    conn.execute(text("""create table transactions (
					  transaction_id BIGINT PRIMARY KEY, 
					  account_id BIGINT,
					  product_id BIGINT,
                      amount DECIMAL(18,2), 
					  transaction_type_id BIGINT,
                      transaction_requested_at BIGINT,
                      transaction_completed_at BIGINT,
                      transaction_status VARCHAR(50),
                      currency VARCHAR(10)
                      )
					  """)) 

    conn.execute(text("""insert into transactions  
        select id, account_id, 2 as product_id, amount, 21 as transaction_type_id,  transaction_requested_at , transaction_completed_at, status , 'BRL' as currency
        from transfer_ins
        union all 
        select id, account_id, 2 as product_id, amount, 22 as transaction_type_id,  transaction_requested_at, transaction_completed_at, status , 'BRL' as currency
        from transfer_outs
        union all 
        select id, account_id, 1 as product_id, pix_amount, case when in_or_out = 'pix_in' then 11 else 12 end as transaction_type_id,  pix_requested_at, pix_completed_at, status , 'BRL' as currency
        from pix_movements""")) 
    
    conn.execute(text("""DROP TABLE pix_movements""")) 
    conn.execute(text("""DROP TABLE transfer_outs""")) 
    conn.execute(text("""DROP TABLE transfer_ins""")) 

    conn.commit()
```
